```python
teams=np.sort(data['Home'].unique())
cartesian_product = [(x, y) for x, y in product(teams, teams) if x != y]
print(cartesian_product)
Tau=[]
for couple in cartesian_product:
    Home=couple[0]
    Away=couple[1]
    matches=data_tot[(data_tot['Home'] == Home) & (data_tot['Away'] == Away)]
    if len(matches)>=10:
        x=[]
        y=[]
        for idx, match in matches.iterrows():
            home_score = match['Hscore']
            away_score = match['Ascore']
            x.append( home_score)
            y.append( away_score)
        tau=stats.kendalltau(y,x)[0]
        Tau.append(tau)
print(Tau)
t=statistics.mean(Tau)
print(t)
```

```python
def kendall(tau):
    def debye_function(x, n):
        def integrand(t):
            return (t**n) / (np.exp(t) - 1)
        
        integral, _ = quad(integrand, 0, x)
        return (n / x**n) * integral
    return(1-4*(1/tau)*(1-debye_function(tau,1)))

def spearman(tau):
    def debye_function(x, n):
        def integrand(t):
            return (t**n) / (np.exp(t) - 1)
        
        integral, _ = quad(integrand, 0, x)
        return (n / x**n) * integral
    return 1-12*(1/tau)*(debye_function(tau,1)-debye_function(tau,2))

    

print(kendall(-0.32),spearman(-0.341))
```
