```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
model_2=pd.read_csv("D:/tipe/tipe-2/csv-files-3/data_glob_0.002.csv")
model_1=pd.read_csv("D:/tipe/tipe-2/csv-files-3/data_glob_model0_0.005.csv")
model_0=pd.read_csv("D:/tipe/tipe-2/csv-files-3/data_glob_model0_0.csv")
```

```python
def bet_1X2(t,row):
    proba_list=[row.proba_win,row.proba_draw,row.proba_lose]
    X2_book=[row.Odds_1,row.Odds_X,row.Odds_2]
    E=[(proba_list[i]*X2_book[i]) for i in range(3)]
    maximum=max(E)
    i=E.index(maximum)
    O=X2_book[i]
    if maximum>t:
        if i==0:
            if row.Hscore>row.Ascore:
                return (10*(O-1),1)
            else :
                return (-10,1)
        elif i==1:
            if row.Hscore==row.Ascore:
                return (10*(O-1),1)
            else :
                return (-10,1)
        else:
            if row.Hscore<row.Ascore:
                return (10*(O-1),1)
            else :
                return (-10,1)
    else:
        return (0,0)
```

```python
def bet_OU(t,row):
    proba_list=[row.proba_O,row.proba_U]
    ou_book=[row.O,row.U]
    E=[(proba_list[i]*ou_book[i]) for i in range(2)]
    maximum=max(E)
    i=E.index(maximum)
    O=ou_book[i]
    if maximum>t:
        if i==0:
            if row.Hscore+row.Ascore>=3:
                return (10*(O-1),1)
            else :
                return (-10,1)
        else:
            if row.Hscore+row.Ascore<=2:
                return (10*(O-1),1)
            else :
                return (-10,1)
    else:
         return (0,0)
```

```python
def X_profit(t,data_all):
    profit=0
    c=0
    for row in [row for row in data_all.itertuples()][237:]:
        profit+=bet_1X2(t,row)[0]
        c+=bet_1X2(t,row)[1]
    return profit,c
def OU_profit(t,data_all):
    c=0
    profit=0
    for row in [row for row in data_all.itertuples()][237:]:
        profit+=bet_OU(t,row)[0]
        c+=bet_OU(t,row)[1]
    return profit,c
```

```python
t=np.linspace(1,2,80)
def results(data):
    X_p=[]
    OU_p=[]
    nb_bets_1X2=[]
    nb_bets_OU=[]
    for i in t:
        _1X2_=X_profit(i,data)
        _OU_=OU_profit(i,data)
        X_p.append(_1X2_[0])
        OU_p.append(_OU_[0])
        nb_bets_1X2.append(_1X2_[1])
        nb_bets_OU.append(_OU_[1])
    return (X_p,OU_p,nb_bets_1X2,nb_bets_OU)
```
