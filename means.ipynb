```python
import pandas as pd
import statistics
import numpy as np
import os 
import glob
```

```python
df1=pd.read_csv("D:/tipe/tipe-2/csv-files-2/2023-2024.csv")
teams=np.sort(df1['Home'].unique())
print(teams)
path="D:\\tipe\\tipe-2\\csv-folders-3\\results_0.002"
A=[]
D=[]
for team in teams :
    deff=[]
    att=[]
    for filename in list(glob.glob(f"{path}/*"))[24:]:
        att_coeff=pd.read_csv(os.path.join(path,filename,f"att_coeff.csv"))
        def_coeff=pd.read_csv(os.path.join(path,filename,f"def_coeff.csv"))
        att.append(att_coeff.loc[att_coeff['teams'] ==team , 'att_coef'].values[0])
        deff.append(def_coeff.loc[def_coeff['teams'] ==team , 'def_coef'].values[0])
    D.append(-statistics.mean(deff))
    A.append(statistics.mean(att))
print(A)
print(D)
A_sorted=sorted(A)
print([20-A_sorted.index(el) for el in A])
D_sorted=sorted(D)
print([20-D_sorted.index(el) for el in D])
```
