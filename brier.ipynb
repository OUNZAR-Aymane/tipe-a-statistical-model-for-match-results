```python
import pandas as pd
teta=0.006
model_2=pd.read_csv("D:/tipe/tipe-2/csv-files-3/data_glob_0.002.csv")
model_1=pd.read_csv("D:/tipe/tipe-2/csv-files-3/data_glob_model0_0.005.csv")
model_0=pd.read_csv("D:/tipe/tipe-2/csv-files-3/data_glob_model0_0.csv")
L=list(model_0.itertuples())[237:]
N=len(L)
print(N)
```

```python
def prediction_1(model):
    w,d,l=0,0,0
    w_r,d_r,l_r=0,0,0
    for row in list(model.itertuples())[237:]:
        L=[row.proba_win,row.proba_draw,row.proba_lose]
        maximum=max(L)
        i_max=L.index(maximum)
        if row.Hscore>row.Ascore:
            w_r+=1
            if i_max==0:
                w+=1
        elif row.Hscore==row.Ascore:
            d_r+=1
            if i_max==1:
                d+=1    
        else:
            l_r+=1
            if i_max==2:
                l+=1  
    return (w_r,w),(d_r,d),(l_r,l)

print(prediction_1(model_0),prediction_1(model_1),prediction_1(model_2))
```

```python
def prediction_2(model):
    o,u=0,0
    o_r,u_r=0,0
    for row in list(model.itertuples())[237:]:
        L=[row.proba_O,row.proba_U]
        maximum=max(L)
        i_max=L.index(maximum)
        if row.Hscore+row.Ascore>2.5:
            o_r+=1
            if i_max==0:
                o+=1
        else:
            u_r+=1
            if i_max==1:
                u+=1  
    return (o_r,o),(u_r,u)

print(prediction_2(model_0),prediction_2(model_1),prediction_2(model_2))
```

```python
def brier_1(model):
    B=0
    for row in list(model.itertuples())[237:]:
        pw=row.proba_win
        pn=row.proba_draw
        pl=row.proba_lose
        if row.Hscore>row.Ascore:
            B+=(pw-1)**2+pn**2+pl**2
        elif row.Hscore<row.Ascore:
            B+=(pl-1)**2+pn**2+pw**2
        else: 
            B+=(pn-1)**2+pw**2+pl**2
    return B/N
print(brier_1(model_0),brier_1(model_1),brier_1(model_2))
```

```python
def brier_2(model):
    B=0
    for row in list(model.itertuples())[237:]:
        po=row.proba_O
        pu=row.proba_U
        if row.Hscore+row.Ascore>2.5:
            B+=(po-1)**2+pu**2
        else: 
            B+=(pu-1)**2+po**2
    return B/N

print(brier_2(model_0),brier_2(model_1),brier_2(model_2))
```
