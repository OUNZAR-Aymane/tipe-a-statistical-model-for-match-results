```python
from bs4 import BeautifulSoup
import os
import requests
import pandas as pd
from datetime import datetime
```

```python
url='https://fbref.com/en/comps/9/2010-2011/schedule/2010-2011-Premier-League-Scores-and-Fixtures' 
page=requests.get(url) 
soup=BeautifulSoup(page.text)
```

```python
Hometeams=[team.text for team in soup.find_all('td', class_='right') if team['data-stat'] == 'home_team' and 'iz' not in team.get('class', [])]
print(soup.find_all('td', class_='right'))
Awayteams=[team.text for team in soup.find_all('td', class_='left') if team['data-stat'] == 'away_team' and 'iz' not in team.get('class', [])]
```

```python
score = [score.text for score in soup.find_all('td', class_='center') if score['data-stat'] == 'score' and 'iz' not in score.get('class', [])]
print(score)
Hscore=[int(score[0]) for score in score]
Ascore=[int(score[2]) for score in score]
```

```python
dates= [datetime.strptime(date.text, '%Y-%m-%d') for date in soup.find_all('td',class_='left') if date['data-stat']=='date'and 'iz' not in date.get('class', [])]
date_diff=[(max(dates) - date).days for date in dates]
print(date_diff)
```

```python
df=pd.DataFrame({'Home':Hometeams,'Hscore':Hscore,'Ascore':Ascore,'Away':Awayteams,'date':dates})
df
df.to_csv(r'D:\\tipe\\tipe-2\\csv-files-2\\2010-2011.csv', encoding='utf-8', index=False)
```
